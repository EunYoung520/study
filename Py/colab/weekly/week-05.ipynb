#인덱싱과 슬라이싱

*    작성일: 2024-10-06
-작성자: 황은영
-주제: 인덱싱과 슬라이싱

```python
import numpy as np
x=np.arange(10)
print(x)
#2번째 원소
print(x[1])
#2,5원소
print(x[[1,4]])
#1~3원소
print(x[0:3])

#1~마지막 원소
print(x[:])
#처음부터 세번째 원소
print(x[:3])
#처음부터 한개씩 건너뛰면서 원소 가져오기
print(x[::2])
```

```python
import numpy as np
matrix=np.arange(1,13,1).reshape(3,4)
print(matrix)
```

```python
#인덱싱
#0행1열의값출력
matrix[0,1]
```

```python
#슬라이싱
#행렬의 일부인 9, 10 출력
print(matrix[2:,:2])
```

```python
print(matrix[:2,1:3])
```

```python
matrix[:,::2]
```

###난수 발생

```python
x=np.random.randint(1,5,size=(10,5))
x
```

```python
print(x[x==1].size)
print(x[x==2].size)
print(x[x==3].size)
print(x[x==4].size)
```

```python
x=np.random.rand(10,5)
x
```

```python
x=np.random.normal(5,3,size=(10,5))#평균 5, #표준편차 3
x, np.mean(x), np.std(x), x.min(), x.max()
```

```python
x=np.random.randn(10,5)#평균 5, #표준편차 1
x, np.mean(x), np.std(x), x.min(), x.max()
```

```python
x=np.array([5,6,16,9,-5,0,7,8,1,10,0,10])
print(len(x))
print(np.count_nonzero(x))
```

```python
x=np.array([5,6,16,9,-5,13,7,8,1,10,0,10])
#평균(mean) numpy 모듈
print(np.mean(x))
#배열 메소드 평균
print(x.mean())
```

```python
x=np.array([5,6,16,9,-5,13,7,8,1,10,0,10])
#분산(var)
print(np.var(x))
print(x.var())
```

```python
x=np.array([5,6,16,9,-5,13,7,8,1,10,0,10])
#표준편차(std)
print(np.std(x))
print(x.std())
```

```python
x=np.array([5,6,16,9,-5,13,7,8,1,10,0,10])
#최댓값
print(max(x))
print(np.max(x))
print(x.max())

#최솟값
print(min(x))
print(np.min(x))
print(x.min())

#중앙값(median)
print(np.median(x))
```
