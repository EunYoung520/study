##시리즈 생성

```python
import pandas as pd
data=pd.Series([1,2,3,4])
data
```

```python
#data 자료형
print(type(data))
#시리즈 (data)의 값 출력
print(data.values)
#시리즈 값의 자료형
print(type(data.values))
```

```python
data=pd.Series([1,2,3,4],dtype='float')
data
```

```python
series=pd.Series([1,2,3,4])
print(series)
#시리즈 생성시 index부여
series=pd.Series([1,2,3,4], index=['a','b','c','d'])
print(series)
#인덱스 ('a')로 접근, 요소 변경(5)
series['a']=5
print=(series)
```

#딕셔너리를 활용하여 Serise 생성

```python
fruits_dict={'사과': 3000, '배': 2000, '수박':10000}
fruits=pd.Series(fruits_dict)
fruits
```

```python
#시리즈 인덱스 확인
print(fruits.index)
```

```python
#시리즈 values(값) 확인
print(fruits_dict)
```
