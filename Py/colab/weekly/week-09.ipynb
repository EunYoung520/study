```python
import numpy as np
import pandas as pd

x=np.arange(10)
fig, ax=plt.subplots()
ax.plot(
    x,x,label='y=x',
    linestyle='-',
    marker='.',
    color='blue'
)
ax.plot(
    x, x**2, label='y=x^2',
    linestyle='-',
    marker='s',
    color='red'
)
fig.savefig('plot.png')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x=np.arange(10)
fig, ax=plt.subplots()
ax.plot(
    x,x,label='y=x',
    linestyle='-',
    marker='.',
    color='blue'
)
ax.plot(
    x, x**2, label='y=x^2',
    linestyle='-',
    marker='s',
    color='red'
)
ax.set_xlabel('x')
ax.set_ylabel('y')

ax.legend(
    loc='best',
    shadow=False,
    fancybox=True,
    borderpad=2
)

fig.savefig('plot.png')
```

## Tips
- 컬럼설명
- total_bill= 총금액
- tip= 팁금액
- sex= 계산한 사람의 성별
- smoker= 흡연여부
- day= 요일
- time= 런치 or 디너
- size= 테이블당 사람 수

```python
tips=sns.load_dataset('tips')
tips
```

```python
sns.countplot(data=tips,x='sex', hue='sex')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',hue='day')
```

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='smoker')
```

```python
sns.catplot(data=tips,x='sex',y='tip',hue='sex')
```
