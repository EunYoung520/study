```python
map_osm=folium.Map(location=[37.5502, 126.982], zoom_start=12)
for region in set(corona_seoul['지역']):
  count=len(corona_seoul[corona_seoul['지역']==region])
  CRS_region=CRS[CRS['시군구명_한글']==region]
  lat=float(CRS_region['위도'])
  lng=float(CRS_region['경도'])
  folium.Marker(location=[lat, lng],
                tooltip=region+"("+str(count)+")").add_to(map_osm)
map_osm
```

###여행력 있다vs없다

```python
df_corona.isnull()
```

```python
sum_travel_no=sum(df_corona['여행력'].isnull())
print(f"여행력이 없는 사람의 수: {sum_travel_no}")
```

```python
sum_travel_all=len(df_corona['여행력'])
print(f"전체 확진자 수: {sum_travel_all}")
sum_travel_yes=sum_travel_all-sum_travel_no
print(f"여행력이 있는 사람의 수: {sum_travel_yes}")
```

```python
plt.rc('font', family='NanumGothic')
plt.ylabel('확진자 수')
plt.xlabel('여행력 여부')
plt.bar(['없다', '있다'], [sum_travel_no, sum_travel_yes])
```
