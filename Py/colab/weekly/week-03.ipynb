###무한루프 & break

```python
s=0
for i in range(1,1000):
  s+=i
  print(i,end=' ')
  if s>100:
    break
print(s)
```

```python
s=0
i=1
while True:
  s+=i
  print(i,end=' ')
  if s>100:
    break
  i+=1
print()
print(s)
```
