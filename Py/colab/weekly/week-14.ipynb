##모델로 성능 검증하기

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
def train_logistic_model(x,y):
  #모델 정의
  model=LogisticRegression()
  #Train/Test셋을 분리
  x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2)
  #학습
  model.fit(x,y)
  #테스트 셋을 예측
  Y_pred=model.predict(x_test)
  print('---'*5+'train_test_split 20/80')
  #스코어 출력
  print('Logistic Regreeion', round(model.score(x_test, y_test)*100, 2))
  print('---'*5)
  return model
```

##데이터 생성

```python
from sklearn.datasets import make_blobs

x,y=make_blobs(n_samples=150, n_features=2, centers=3,
               cluster_std=0.5, shuffle=True, random_state=0)
print(f"x shape: {x.shape}, y shape: {y.shape}")
```

```python
type(y)
```

```python
#초기 데이터 시각화
import matplotlib.pyplot as plt
plt.scatter(x[:,0], x[:,1], marker='o', s=30)
plt.grid()
plt.show()
```

##Kmeans Clustering

```python
from sklearn.cluster import KMeans
init_centers="random"
```

```python
kmodel=KMeans(n_clusters=3, init=init_centers, random_state=3)
pred=kmodel.fit_predict(x)
```

```python
x
```

```python
x[pred==1]
```

```python
plt.scatter(x[pred==0][:,0],x[pred==0][:,1], marker='o', s=50, c='yellow', label='cluster1')
plt.scatter(x[pred==1][:,0],x[pred==1][:,1], marker='s', s=50, c='green', label='cluster2')
plt.scatter(x[pred==2][:,0],x[pred==2][:,1], marker='v', s=50, c='blue', label='cluster3')
plt.scatter(kmodel.cluster_centers_[:,0], kmodel.cluster_centers_[:,1],
            marker='*', s=80, c='red', label='center')
plt.legend()
plt.grid()
plt.show()
```

##n-Clusters를 알기 위한 방법

```python
#엘보우 기법
def elbow(x):
  sse=[]
  for i in range(1,11):
    km=KMeans(n_clusters=i, init='k-means++', random_state=0)
    km.fit(x)
    sse.append(km.inertia_)
  print('-'*20)
  print(sse[3])
  plt.plot(range(1,11), sse, marker='o')
  plt.axvline(x=3, color="r")
  plt.axhline(y=sse[3], color="r")
  plt.xlabel('Number of clusters')
  plt.ylabel('ssd')
  plt.show()
elbow(x)
```
