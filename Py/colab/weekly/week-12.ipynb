```python
#데이터 생성(50개 생성)
import numpy as np
import matplotlib.pyplot as plt
x=5*np.random.rand(50)
y=3*x+np.random.rand(50)
plt.scatter(x,y)
```

```python
#모델선택
from sklearn.linear_model import LinearRegression
```

```python
#모델의 하이퍼파라미터 선택
model=LinearRegression(fit_intercept=True) #y절편 구하기
model
```

```python
#데이터를 특징 배열과 대상 벡터로 배치, x(1개의 열을 가진 2차원 배열), y(벡터)
#차원확인
x.shape
```

```python
x.reshape(2,-1)
```

```python
x
```

```python
X=x.reshape(-1,1)
X.shape
```

```python
X
```

```python
#모델에 데이터 적합
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
#new데이터에 적용
xfit=np.linspace(-1,11).reshape(-1,1)
yfit=model.predict(xfit)
```

```python
plt.scatter(x,y)
plt.plot(xfit,yfit,'--r')
```

##샘플 데이터 예시

```python
#feature
x=pd.DataFrame(house.data, columns=house.feature_names)
x.head()
```

```python
y=pd.DataFrame(house.target, columns=house.target_names)
y.head()
```

```python
data=pd.concat([x,y], axis=1)
data.head()
```

```python
data.describe()
```

```python
x=data.drop('MedHouseVal', axis=1)
```

```python
y=data['MedHouseVal']
```

```python
#데이터 프레임에서 target과 feature분리
x=data.drop('MedHouseVal', axis=1)
y=data['MedHouseVal']
```

```python
x.head()
```

```python
y.head()
```

```python
#학습데이터와 테스트용 데이터 분리
from sklearn.model_selection import train_test_split
xTrain, xTest, yTrain, yTest=train_test_split(x,y, test_size=0.2, random_state=123)
```

```python
xTrain
```

```python
len(xTest)/len(x)
```

##지도학습_회귀

```python
from sklearn.linear_model import LinearRegression
```

```python
#모델에 데이터 적합시키기
model=LinearRegression()
model.fit(xTrain, yTrain)
```

```python
#예측하기
y_pred=model.predict(xTest)
```

```python
#평가하기

from sklearn.metrics import r2_score
m_R2=r2_score(yTest, y_pred)

print(m_R2)
```

```python
y_pred2=model.predict(xTrain)
m2_R2=r2_score(yTrain, y_pred2)
print(m2_R2)
```
