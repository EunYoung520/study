###사용자 정의 함수

```python
#선의 길이를 인수로 받아 선을 그어주는 함수 정의
#매개변수를 이용해서 함수 내부로 값을 전달
def line2(width):
  print('-'*width)
```

```python
#매개변수로 선의 길이를 넘겨붐
#함수 호출 시 선길이를 인수로 넣어줌
line2(30)
line2(50)
```

###가변인자

```python
def sum_nums(*numbers):
  t=0
  for i in numbers:
    t+=i
  return t
```

```python
print(sum_nums(1,2,3))
print(sum_nums(10,20,30,40,50,60))
```

##각의 수를 입력받아 내각과 외각의 크기를 리턴하는 함수 작성

```python
def shape_angle(n):
  in_=(180*(n-2))/n
  out_=360/n
  return in_,out_
```

```python
n=int(input('삼각형:3, 사각형:4...:'))
in_, out_=shape_angle(n)
print(f'내각: {in_}, 외각: {out_}')
```
