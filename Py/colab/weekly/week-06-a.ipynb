```python
#딕셔너리를 활용해 데이터 프레임(country) 생성
data={'country': ['kr', 'ja','ch','usa'],
      'gdp':[16931, 51670, 140925, 204128],
      'population':[5180, 12718, 141500, 32676]}
country=pd.DataFrame(data) #데이터 프레임 생성
country
```

##조건을 충족하는 값 추출

```python
import numpy as np
import pandas as pd
df=pd.DataFrame(np.random.rand(5,2),columns=['A','B'])
print(df)
```

```python
df['A']<0.5
```

```python
df['B']>0.3
```

```python
(df['A']<0.5)&(df['B']>0.3)
```

```python
'''
A컬럼값이 0.5보다 작고
B컬럼값이 0.3보다 큰값 추출
'''
df[(df['A']<0.5)&(df['B']>0.3)]
```

```python
df.query('A<0.5 and B>0.3')
```

```python
df
```

```python
'''
A컬럼의 값이 0.6보다 크거나
B컬럼의 값이 0.6보다 큰 행 추출
'''
df[(df['A']>0.6) | (df['B']>0.6)]
```

```python
df.query('A>0.6 or B>0.6')
```
