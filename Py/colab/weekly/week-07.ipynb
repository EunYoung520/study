##map
-    반복 가능한 iterable객체를 받아서, 각 요소에 함수를 적용해주는 함수
```
t=[1,2,3,4,5]
```
위와 같은 iterable객체인 list의 값들에 각각 +1을 하고 싶은 경우
```
#적용할 요소들
[1,2,3,4]
#적용시킬 연산
def add_1(n):
  return n+1
```



````python
def plus_1(n):
  return n+1

t=[1,2,3,4]
result=[]
for i in t:
  result.append(plus_1(i))

result
````

````python
def pius_1(n):
  return n+1
#map(적용할 함수, 적용할 요소들)
t=[1,2,3,4]
result=map(pius_1, t)
print(list(result))
````

````python
t=[1,2,3,4]
result=map(lambda x:x+1, t)
print(list(result))
````

##filter
-    filter함수는 특정 조건으로 걸러서 걸러진 요소들로 iterator객체를 만들어서 리턴
-    map함수와 사용법은 동일하나, 함수의 결과가 참인지 거짓인지에 따라 해당 요소를 포함할지를 결정
-    숫자를 가지고있는 iterable객체에서 짝수들만 포함하는 객체를 만들 때
```
#적용할 요소들
[1,2,3,...,10]
#적용시킬 함수
def is_even(x):
  return True if x%2==0 else False
```



````python
t=list(range(1,11))
result=[]
def in_even(x):
  return True if x%2==0 else False
print(t)
for i in t:
  if in_even(i):
    result.append(i)

print(result)
````

````python
#filter(적용시킬 함수, 적용할 요소들)
t=list(range(1,11))

def in_even(x):
  return True if x%2==0 else False

result=filter(in_even, t)

print(list(result))
````

````python
t=list(range(1,11))

result=filter(lambda x:x%2==0, t)

print(list(result))
````

````python
import numpy as np
import pandas as pd

print("DataFrame: ")
df=pd.DataFrame({
    'col1':[2,1,9,8,7,4],
    'col2':['A','A','B',np.nan,'D','C'],
    'col3':[0,1,9,4,2,3]
},index=[1,3,2,5,0,4])
print(df,"\n")
````

````python
#index값 기준으로 정렬하기
df=df.sort_index(axis=0,ascending=False)
````

````python
#열 인덱스 기준으로 정렬하기
df=df.sort_index(axis=1,ascending=False)
df
````

````python
df=df.sort_index(axis=0,ascending=True)
````

````python
df
````

````python
#정렬 코드 입력해보기
#1.col1을 기준으로 오름차순으로 정렬하기
print('col을 기준 오름차순')
sorted_df1=df.sort_values('col1',ascending=True)
print(sorted_df1)
print(df)
````

````python
#2.col2를 기준으로 내림차순으로 정렬하기

sortes_df2=df.sort_values('col2',ascending=True)
print(sortes_df2)
````

````python
#3,col2를 기준으로 오름차순으로 col1를 기준으로 내림차순으로 정렬하기
sorted_df3=df.sort_values(['col2', 'col1'], ascending=[True, False])
print(sorted_df3)
````

##그룹으로 묶기(1)
>groupby()함수를 이용하면 키 값을 기준으로 그룹으로 묶을 수 있다.

>groupby()함수의 사용 예시를 들자면, 시험 성적 데이터에서 각 반을 그룹으로 묶어 각 반의 평균 점수를 구할 수 있습니다.

>그룹으로 묶는다는 의미가 잘 와닿지 않으신가요?

>이렇게 해보세요!
1.    데이터 프레임의 key칼럼을 groupby함수로 묶고, key별로 data1과 data2 합게를 출력해보세요
2.    데이터 프레임의 key와 data1칼럼을 groupby함수로 묶고, key와 data1별로 data2의 합계를 출력해보세요.

````python
import numpy as np
import pandas as pd

df=pd.DataFrame({
    'key':['A','B','C','A','B','C'],
    'data1':[1,2,3,1,2,3],
    'data2':[4,4,6,0,6,1]
})
print("DataFrame:")
print(df,"\n")

#groupby함수를 이용해봅시다
#key를 기준으로 묶어 합계를 구해 출력해보세요.
print(df.groupby('key').sum())

#key와 data1을 기준으로 묶어 합계를 구해 출력해보세요.
print(df.groupby(['key','data1']).sum())
````
