```python
#라이브러리 모듈 불러오기

#pandas 모듈을 pd라는 별칭으로 불러오기
#numpy 모듈을 np라는 별칭으로 불러오기
import pandas as pd
import numpy as np
```

```python
#list를 활용하여 데이터프레임 생성하시오
student_data=[['홍길동',80,70,80,],
              ['박철수', 60, 70, 80],
              ['김영희', 70, 90,80]]
df=pd.DataFrame(student_data)
df
```

```python
#columns옵션을 주어 컬럼 이름을 지정하여 데이터프레임을 생성하라
df = pd.DataFrame(student_data, columns=['이름','국어','수학','영어'])
df
```

```python
#데이터 프레임에서 '국어'칼럼을 추출하라
df['국어']
```

```python
#iloc[]을 이용하여 '홍길동'행(0번째 인덱스)를 조회하라
df.iloc[0]
```

```python
#사회 컬럼을 데이터프레임에 추가하라
df['사회']=[50,60,70]
df
```

```python
#'사회' 칼럼을 삭제하라
df=df.drop('사회',axis=1)
df
```

```python
#df.loc[행,열]명령어를 사용하여 '박철수'가 출력되도록 하라
rest=df.loc[1,'이름']
rest
```

```python
#박철수의 영어를 100으로 변경하라
df.loc[1,'영어']=100
df
```

```python
#국어가 70 이상인 데이터만 출력하라
df[df['국어']>=70]
```

```python
#국어가 70이고, 수학이 80이 넘는 데이터만 추출하라
df[(df['국어']==70) & (df['수학']>=80)]
```

```python
#국어가 80 이상이거나, 영어가 80 이상인 데이터만 추출하라
df[(df['국어']>=80) | (df['영어']>=80)]
```

```python
#set_index('컬럼명')을 사용하여 이름 칼럼을 행 인덱스로 적용하라
df = df.set_index('이름')
df
```

```python
#홍길동의 국어를 조회하라
df.loc['홍길동', '국어']
```

```python
#열 방향으로 합계를 계산하라
df.sum(axis=0)
```

```python
#행 방향으로 합계를 계산하라
df.sum(axis=1)
```

```python
#행 방향으로 합계를 구해 '합계'라는 새로운 칼럼을 추가하라
df['합계'] = df.sum(axis=1)
df
```

```python
#본인의 성적을 추가하라
df.loc['정은미'] = [100, 85, 75, float('nan')]
df
```

```python
#결측값을 확인하라
df.isnull()
```

```python
#결측값의 개수를 출력하라
df.isnull().sum()
```

```python
#'합계' 컬럼을 '총점'컬럼으로 변경하라
df.rename(columns={'합계':'총점'}, inplace=True)
df
```

```python
#정은미의 합계를 구해서 업데이트 하라
df.loc['정은미','총점']=df.loc['정은미'].sum()
df
```

```python
#다음 조건에 해당하는 데이터를 필터링하라
'''
조건:
age와 name열만 출력
age가 60세 이상
상위 3개만 출력
'''
titanic_df.loc[titanic_df['Age'] >= 60, ['Age', 'Name']].head(3)
```

```python
'''
조건:
name, pclass, age열만 출력
survived가 1인 경우
상위 5개만 출력
'''
titanic_df.loc[titanic_df['Survived'] == 1, ['Name', 'Pclass', 'Age']].head()
```

```python
#다음 조건에 해당하는 데이터의 개수를 출력하시오

'''
조건:
age는 60세 이상
pclass는 1등급
sex가 female
'''
titanic_df.loc[(titanic_df['Age'] >= 60) & (titanic_df['Pclass'] == 1) & (titanic_df['Sex'] == 'female')].shape[0]
```

```python
#Pclass별 felame 생존자 수를 출력하라
female_survivors = titanic_df[(titanic_df['Sex'] == 'female') & (titanic_df['Survived'] == 1)]
survivor_counts = female_survivors.groupby('Pclass')['Survived'].count()
for pclass, count in survivor_counts.items():
    print(f"{pclass}등급 여성 생존자 수 {count}")
```

```python
#Pclass를 인덱스로 하여 각 컬럼의 개수를 구하라
titanic_df.groupby('Pclass').count()
```

```python
#제공된 iris.csv파일을 불러와 iris에 할당하라
iris = pd.read_csv('iris.csv')
iris
```

```python
#class칼럼의 고유한 데이터가 어떤 것들이 있는지 확인하라
iris['class'].unique()
```

```python
#class 칼럼의 고유한 데이터가 몇 개인지 확인하라
len(iris['class'].unique())
```

```python
#class컬럼의 데이터가 각 개체별로 몇 개씩 있는지 확인하라
iris['class'].value_counts()
```

```python
#iris의 꽃받침(sepal) 길이만 상위 10개를 살펴보라
iris['sepal length'].head(10)
```

```python
#iris의 꽃받침의 5행~19행까지 sclicing하라
iris.loc[5:19, ['sepal length']]
```

```python
#컬럼별로 일괄 계산이 가능하다. 'sepal length'컬럼을 'petal length'컬럼으로 나눈 값을 출력하라
iris['sepal length'] / iris['petal length']
```

```python
#sepal length가 5보다 큰 값만 추출하라
iris[iris['sepal length'] > 5]
```

```python
#sepal length가 5보다 크거나, petal length가 1.5보다 작은 데이터만 추출하라
iris[(iris['sepal length'] > 5) | (iris['petal length'] < 1.5)]
```

```python
#sepal length가 5보다 크면서 petal length가 1.5보다 작은 데이터만 추출하라
iris[(iris['sepal length'] > 5) & (iris['petal length'] < 1.5)]
```

```python
#sepal length>5이고 petal length<1.5 인 petal width열만 추출하라
iris.loc[(iris['sepal length'] > 5) & (iris['petal length'] < 1.5), 'petal width']
```

```python
#class별로 각 컬럼의 합을 출력하라 grupby()사용
iris.groupby('class').sum()
```

```python
#class별로 각 컬럼의 가장 큰 값과 가장 작은 값의 차이를 출력하라
iris.groupby('class').agg(lambda x: x.max() - x.min())
```
